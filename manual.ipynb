## **Thứ tự ưu tiên chọn locator**

1. getByRole     → ưu tiên nhất, theo chuẩn accessibility
2. getByLabel    → cho input có label
3. getByText     → cho text hiển thị
4. getByTestId   → nếu dev đặt data-testid
5. locator('#id') → dùng id
6. locator('.class') → dùng class, hạn chế dùng

## **Công thức nhìn HTML → chọn locator**

Nhìn vào tag trước:
─────────────────────────────────────────
```

<button>                        → getByRole('button', { name: '...' })
<a>                             → getByRole('link', { name: '...' })
<input type=text>               → getByLabel('...') hoặc locator('#id')
<input type=submit>             → toHaveAttribute('value', '...')
<li>, <div>, <span> có text     → getByText('...')

```


## **Công thức nhìn attribute → chọn locator**

Có id="..."             → locator('#id')        ← nhanh, chắc

Có text rõ ràng         → getByText / getByRole

Có data-testid          → getByTestId

Chỉ có class            → hạn chế dùng, class hay thay đổi




## **Checklist F12 nhanh**
Mỗi lần inspect element, tự hỏi theo thứ tự:

□ Tag là gì?        (button, input, a, div...)

□ Có id không?      → dùng #id

□ Có text không?    → dùng getByText hoặc getByRole

□ Có role không?    → dùng getByRole

□ Có data-testid?   → dùng getByTestId

□ Chỉ có class?     → tìm cách khác, class không ổn định


## **Ví dụ thực tế từ case trước**

``` 
html
<label for="username">Benutzername</label>
→ tag: label, có for="username"
→ dùng: locator('label[for="username"]')

<input id="username" />
→ tag: input, có id
→ dùng: locator('#username')

<input type="submit" value="Anmelden" />
→ tag: input type submit, không có text
→ dùng: locator('#kc-login') + toHaveAttribute('value', '...')

<a role="menuitem">German (Deutsch)</a>
→ tag: a, có role, có text
→ dùng: getByRole('menuitem', { name: 'Deutsch', exact: false })
```
